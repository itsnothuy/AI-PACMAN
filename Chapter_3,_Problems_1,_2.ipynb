This notebook shows how to solve Problems 1 and 2 from Chapter 3.

```python
import numpy as np
```

Start by defining the matrix $P$ and the invariant distribution $\pi$.

```python
P = np.matrix([[0.3, 0.2, 0.5], [0.4, 0.3, 0.3], [0.3, 0.4, 0.3]]).astype(np.float64)
P
```

```python
pi = np.matrix([37/112, 34/112, 41/112]).astype(np.float64)
pi
```

Let's use the built-in norm function in NumPy.

```python
from numpy.linalg import norm
```

```python
v = np.matrix([1, 0, 0])
```

```python
norm(v-pi)
```

```python
def diff(t, v0 = np.matrix([1, 0, 0]).astype(np.float64)):
  return norm(v0*P**t-pi)
```

```python
v0 = np.matrix([1/2, 1/3, 1/6]).astype(np.float64)
for t in range(20):
    print(diff(t, v0 = v0))
```

```python
def diff_ratio(t, v0 = np.matrix([1, 0, 0])):
  return diff(t+1, v0 = v0)/diff(t, v0 = v0)
```

```python
v0 = np.matrix([1/2, 1/3, 1/6])
for t in range(20):
    print(diff_ratio(t, v0 = v0))
```

```python
diff_ratios = np.array([diff_ratio(t, v0=v0) for t in range(18)])
diff_ratios.mean()
```

```python
evals = np.linalg.eigvals(P.T)
for ev in evals:
  print(np.abs(ev))
```

A standard result says that $|p_t-\pi|\leq C\cdot|\lambda_2|^t$, where $C$ is a fixed constant, independent of $t$ (which can actually be expressed explicitly in terms of the eigenvalues of $P$).  If we assume $|p_t-\pi|\approx C\cdot|\lambda_2|^t$, dividing gives $\frac{\mathrm{diff}(t+1)}{\mathrm{diff}(t)}\approx|\lambda_2|$.  We see this happen above until we run out of numerical precision.
